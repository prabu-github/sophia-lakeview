```python
from pathlib import Path
import numpy as np
import pandas as pd
```

```python
compatible_dir = Path('io/data/compatible')
csv_file = compatible_dir/'srp.csv'
df = pd.read_csv(csv_file)
print(df.shape)
print()
cols_800 = [c for c in df.columns if ((c[0] == 'X') and (float(c.split('_')[-1]) < 800.01))]
values = df[cols_800]
max_values = np.max(values, axis=1)
print(max_values)
print(f'{np.min(max_values) = }')

negs = (values < 0.0)
print(negs.shape)

all_negs = np.all(negs, axis=1)
for r in range(all_negs.shape[0]):
    if all_negs[r]:
        print(f'All negatives on row: {r}')
```
